```python
def show_balance(balance):
    print(f"잔액은 {balance}원입니다.")

def deposit(balance):
    amount = int(input("금액을 입력하시오: "))
    balance += amount
    print(f"잔액은 {balance}원입니다.")
    return balance

def withdraw(balance):
    amount = int(input("금액을 입력하시오: "))
    if balance >= amount:
        balance -= amount
        print(f"잔액은 {balance}원입니다.")
    else:
        print("잔액이 부족합니다.")
    return balance

pin = input("비밀 번호를 입력하시오: ")

if pin == "4567":
    balance = 0
    while True:
        print("\n1 - 잔액 보기    2- 입금    3- 출금    4- 종료")
        choice = int(input("번호를 선택하시오: "))

        if choice == 1:
            show_balance(balance)
        elif choice == 2:
            balance = deposit(balance)
        elif choice == 3:
            balance = withdraw(balance)
        elif choice == 4:
            break
```
