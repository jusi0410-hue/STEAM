```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from PIL import Image
import requests
from io import BytesIO

url = 'https://cdn.asiaa.co.kr/news/photo/201906/517_1489_1800.jpg'

response = requests.get(url)
pic = Image.open(BytesIO(response.content))

pic_arr = np.asarray(pic)
pic_copy = pic_arr.copy()

pic_red = pic_arr.copy()
pic_red[:,:,1] = 0
pic_red[:,:,2] = 0

pic_green = pic_arr.copy()
pic_green[:,:,0] = 0
pic_green[:,:,2] = 0

pic_blue = pic_arr.copy()
pic_blue[:,:,0] = 0
pic_blue[:,:,1] = 0

pic_combined = pic_red + pic_green + pic_blue

plt.imshow(pic_red)
plt.show()

plt.imshow(pic_green)
plt.show()

plt.imshow(pic_blue)
plt.show()

plt.imshow(pic_combined)
plt.show()
```
