```python
import random

def getLotto():
    myList = []
    while len(myList) != 6:
        number = random.randint(1, 45)
        if number not in myList:
            myList.append(number)
    myList.sort()
    return myList

print("이번 주 예상 로또 번호 10개 조합:")
for i in range(10):
    lottoList = getLotto()
    print(f"{i+1}회: {lottoList}")
```
